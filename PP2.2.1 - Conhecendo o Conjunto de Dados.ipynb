```python
# Item 1.1 — Separando os atributos preditores (todas as colunas, exceto o alvo Class)
preditores = df.columns.drop("Class")
print(f"{len(preditores)} atributos preditores")
```

```python
# Item 1 — Histogramas: distribuição e tendência central (média e mediana) de cada preditor
fig, axes = plt.subplots(4, 4, figsize=(16, 12))

for atributo, ax in zip(preditores, axes.flat):
    ax.hist(df[atributo], bins=50)
    ax.axvline(df[atributo].mean(), color="red", linestyle="--", label="Média")
    ax.axvline(df[atributo].median(), color="green", label="Mediana")
    ax.set_title(atributo)

axes.flat[0].legend()
plt.tight_layout()
plt.show()
```

```python
# Item 1 — Boxplots: dispersão e outliers de cada preditor
fig, axes = plt.subplots(4, 4, figsize=(16, 12))

for atributo, ax in zip(preditores, axes.flat):
    ax.boxplot(df[atributo], orientation="horizontal")
    ax.set_title(atributo)
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```

```python
# Item 1.2 — Medidas estatísticas dos atributos preditores
df[preditores].describe().T
```

```python
# Item 2 — Balanceamento do atributo-alvo: quantidade de exemplos por classe
contagem = df["Class"].value_counts()

contagem.plot(kind="bar")
plt.title("Quantidade de exemplos por classe")
plt.ylabel("Quantidade")
plt.show()
```

```python
# Item 2 — Balanceamento do atributo-alvo: porcentagem de exemplos por classe
(df["Class"].value_counts(normalize=True) * 100).round(2)
```

### Respostas — Visualizando a base de dados

1. A base possui 17 colunas, mas apenas 16 são atributos preditores: a coluna `Class` é o atributo-alvo e, por isso, foi excluída dos gráficos e das medidas estatísticas.

2. O atributo-alvo é desbalanceado. A classe Dermason concentra 26,05% dos exemplos, enquanto a Bombay possui apenas 3,84%, cerca de 7 vezes menos. As demais classes ficam entre 9,71% (Barbunya) e 19,37% (Sira).
