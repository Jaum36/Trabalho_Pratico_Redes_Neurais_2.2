## Busca em Grade

Uma maneira padrão de escolher os parâmetros de um modelo de Machine Learning é por meio de uma busca em grade via força bruta. O algoritmo da busca em grade é dado como segue:

1. Escolha a métrica de desempenho que você deseja maximizar (por exemplo, o F-Score com média macro, adequado a um problema multiclasse com classes desbalanceadas)  
2. Escolha o algoritmo de Machine Learning (exemplo: redes neurais artificiais). Em seguida, defina os parâmetros ou hiperparâmetros deste tipo de modelo sobre os quais você deseja otimizar (número de épocas, taxa de aprendizado, etc.) e construa um array de valores a serem testados para cada parâmetro ou hiperparâmetro.  
3. Defina a grade de busca, a qual é dada como o produto cartesiano dos valores de cada parâmetro a ser testado. Por exemplo, para os arrays [50, 100, 1000] e [10, 15], tem-se que a grade é [(50,10), (50,15), (100,10), (100,15), (1000,10), (1000,15)].
4. Para cada combinação de parâmetros a serem otimizados, utilize _apenas o conjunto de treinamento_ para realizar uma validação cruzada (holdout ou k-fold) e calcule a métrica de avaliação nos conjuntos de validação gerados por ela. O conjunto de teste não participa da busca. Vamos efetuar validação cruzada holdout 70/30 estratificada.
5. Escolha a combinação de parâmetros que maximiza a métrica de avaliação na validação. Este é o modelo otimizado, que deve então ser reavaliado uma única vez no conjunto de teste.

Por que esta abordagem funciona? Porque a busca em grade efetua uma pesquisa extensiva sobre as possíveis combinações de valores para cada um dos parâmetros a serem ajustados. Para cada combinação, ela estima a performance do modelo em dados não utilizados no seu treinamento (os conjuntos de validação). Por fim, o modelo com melhor métrica de desempenho é escolhido. Tem-se então que este modelo é o que tem maior chance de generalizar bem mediante dados nunca antes vistos, e o conjunto de teste, que não foi utilizado na busca, fornece a estimativa final e não enviesada desse desempenho.

## Validação Cruzada k-fold

Na elaboração da busca em grade, vamos avaliar os modelos propostos segundo uma estratégia de validação cruzada ainda não explorada até o momento: a validação cruzada *k-fold*. Segundo a mesma, o conjunto de dados é particionado em *k* partes (*folds*): a cada iteração, separa-se uma das partes para validação e o modelo é treinado com as *k-1* partes remanescentes. Valores sugestivos de *k* na literatura são *k* = 3, 5 ou 10, pois o custo computacional desta validação dos modelos é alto. A métrica de desempenho é resultante da média dos desempenhos nas *k* iterações. A figura a seguir ilustra a ideia desta avaliação:

<img src="https://ethen8181.github.io/machine-learning/model_selection/img/kfolds.png" width="600">

Considerando a métrica de desempenho F1-Score (com média macro, dado que o problema possui 7 classes), considere a validação cruzada 5-fold estratificada para aferir os resultados da busca em grade anterior, realizada *apenas com o conjunto de treinamento*. O conjunto de teste deve ser utilizado uma única vez, para avaliar a melhor configuração encontrada.

```python
## Importação das bibliotecas e preparação dos dados

import warnings
import pandas as pd
import matplotlib.pyplot as plt


from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    GridSearchCV,
    ParameterGrid,
)
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.exceptions import ConvergenceWarning

df = pd.read_excel("Dry_Bean_Dataset.xlsx")
X = df.drop(columns=["Class"])
y = df["Class"]

# Holdout 70/30 estratificado — o teste fica guardado até o final
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, stratify=y, random_state=42
)

# Escalonamento ajustado apenas com o treino
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Treinamento:", X_train_scaled.shape)
print("Teste:", X_test_scaled.shape)
```

```python
# Top-6 RNAs identificadas no PP2.2.2

top6 = pd.DataFrame([
    {"Neurônios por camada": (50, 25), "Ativação": "tanh", "Estratégia": "Própria",
     "F1 macro (média ± desvio)": "0.9422 ± 0.0039", "Acurácia (média ± desvio)": "0.9311 ± 0.0047", "Repetições": 10},
    {"Neurônios por camada": (50,),    "Ativação": "relu", "Estratégia": "Própria",
     "F1 macro (média ± desvio)": "0.9422 ± 0.0038", "Acurácia (média ± desvio)": "0.9308 ± 0.0048", "Repetições": 10},
    {"Neurônios por camada": (14, 18), "Ativação": "tanh", "Estratégia": "Pirâmide Geométrica",
     "F1 macro (média ± desvio)": "0.9418 ± 0.0027", "Acurácia (média ± desvio)": "0.9311 ± 0.0031", "Repetições": 10},
    {"Neurônios por camada": (20,),    "Ativação": "relu", "Estratégia": "Própria",
     "F1 macro (média ± desvio)": "0.9417 ± 0.0038", "Acurácia (média ± desvio)": "0.9309 ± 0.0041", "Repetições": 10},
    {"Neurônios por camada": (19, 13), "Ativação": "tanh", "Estratégia": "Pirâmide Geométrica",
     "F1 macro (média ± desvio)": "0.9417 ± 0.0041", "Acurácia (média ± desvio)": "0.9306 ± 0.0048", "Repetições": 10},
    {"Neurônios por camada": (20,),    "Ativação": "tanh", "Estratégia": "Própria",
     "F1 macro (média ± desvio)": "0.9416 ± 0.0033", "Acurácia (média ± desvio)": "0.9309 ± 0.0041", "Repetições": 10},
])
top6.index = range(1, 7)
top6
```

```python
# Grade de hiperparâmetros (A–E do enunciado)

hiperparametros = {
    "solver": ["adam", "sgd"],                # A. Solver (sem LBFGS)
    "batch_size": [64, 200],                  # B. Batch size
    "learning_rate_init": [0.001, 0.01],      # C. Taxa de aprendizado inicial
    "n_iter_no_change": [10, 20],             # D. Paciência
    "max_iter": [150, 200],                   # E. Épocas
}

# Cada arquitetura mantém sua própria função de ativação
param_grid = [
    {"hidden_layer_sizes": [camadas], "activation": [ativacao], **hiperparametros}
    for camadas, ativacao in zip(top6["Neurônios por camada"], top6["Ativação"])
]

n_candidatos = len(ParameterGrid(param_grid))
print(f"Candidatos na grade: {n_candidatos}")
print(f"Treinamentos com 5 folds: {n_candidatos * 5}")
```

```python
# Busca em grade com validação cruzada 5-fold estratificada (apenas o treino)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

busca = GridSearchCV(
    estimator=MLPClassifier(random_state=42),
    param_grid=param_grid,
    scoring="f1_macro",
    cv=cv,
    n_jobs=-1,
    verbose=1,
)

with warnings.catch_warnings():
    warnings.simplefilter("ignore", ConvergenceWarning)
    busca.fit(X_train_scaled, y_train)

print("Melhores hiperparâmetros:", busca.best_params_)
print(f"Melhor F1 macro médio na validação: {busca.best_score_:.4f}")

resultados = pd.DataFrame(busca.cv_results_)
resultados.sort_values("rank_test_score")[
    ["params", "mean_test_score", "std_test_score"]
].head(10)
```

## Identificando a mellhor solução

Como resultado da busca em grade com validação cruzada 5-fold, identifique o modelo otimizado com melhor desempenho para o problema. Apresente claramente este modelo, sua arquitetura (número de camadas ocultas, neurônios por camada e função de ativação), os hiperparâmetros otimizados (solver, batch size, taxa de aprendizado inicial, paciência e épocas) e os resultados (F1-Score macro) para cada um dos folds avaliados, junto com a média ± desvio padrão entre os folds. Em seguida, reavalie este modelo uma única vez no conjunto de teste, apresentando a matriz de confusão, a acurácia e o F1-Score macro. Esta é a melhor solução identificada em decorrência deste projeto.

```python

# Identificando o melhor modelo e seus resultados na validação cruzada

# Índice da melhor configuração encontrada
indice_melhor = busca.best_index_

# Recuperando o F1-score de cada fold
resultados_folds = [
    busca.cv_results_[f"split{i}_test_score"][indice_melhor]
    for i in range(cv.get_n_splits())
]

# Média e desvio padrão dos resultados
media_f1 = busca.cv_results_["mean_test_score"][indice_melhor]
desvio_f1 = busca.cv_results_["std_test_score"][indice_melhor]

# Recuperando a melhor arquitetura
melhores_parametros = busca.best_params_
camadas = melhores_parametros["hidden_layer_sizes"]

print("MELHOR MODELO ENCONTRADO")
print("-" * 40)

print(f"Camadas ocultas: {len(camadas)}")
print(f"Neurônios por camada: {camadas}")
print(f"Função de ativação: {melhores_parametros['activation']}")

print("\nHIPERPARÂMETROS OTIMIZADOS")
print(f"Solver: {melhores_parametros['solver']}")
print(f"Batch size: {melhores_parametros['batch_size']}")
print(f"Learning rate inicial: {melhores_parametros['learning_rate_init']}")
print(f"Paciência: {melhores_parametros['n_iter_no_change']}")
print(f"Épocas máximas: {melhores_parametros['max_iter']}")

print("\nRESULTADOS DA VALIDAÇÃO CRUZADA")
for i, resultado in enumerate(resultados_folds, start=1):
    print(f"Fold {i}: {resultado:.4f}")

print(f"\nF1-score médio: {media_f1:.4f}")
print(f"Desvio padrão: {desvio_f1:.4f}")
print(f"Resultado: {media_f1:.4f} ± {desvio_f1:.4f}")
```

```python

# Avaliando a melhor configuração no conjunto de teste

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# Recuperando o melhor modelo já ajustado pelo GridSearchCV
melhor_modelo = busca.best_estimator_

# Realizando a previsão no conjunto de teste
y_pred_teste = melhor_modelo.predict(X_test_scaled)

# Calculando as métricas
acuracia_teste = accuracy_score(y_test, y_pred_teste)

f1_teste = f1_score(
    y_test,
    y_pred_teste,
    average="macro"
)

print("RESULTADOS NO CONJUNTO DE TESTE")
print("-" * 40)
print(f"Acurácia: {acuracia_teste:.4f}")
print(f"F1-score macro: {f1_teste:.4f}")

# Construindo a matriz de confusão
matriz = confusion_matrix(
    y_test,
    y_pred_teste,
    labels=melhor_modelo.classes_
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=matriz,
    display_labels=melhor_modelo.classes_
)

disp.plot(cmap="Blues", xticks_rotation=45)
plt.title("Matriz de Confusão — Melhor MLP")
plt.tight_layout()
plt.show()
```

2. Treine o modelo com todos os dados, utilizando a arquitetura e os hiperparâmetros otimizados. O escalonador (`StandardScaler`) deve ser ajustado com todos os dados e salvo junto com a rede, preferencialmente em um único `Pipeline`  
3. Salve o modelo em disco  
4. Construa uma rotina que recupere o modelo em disco  
5. Mostre que a rotina é funcional, fazendo previsões com todos os elementos do dataset e exibindo uma matriz de confusão das mesmas. Como o modelo foi treinado com todos os dados, esta matriz serve apenas para verificar o funcionamento da rotina e não como estimativa de desempenho, que é dada pelos resultados da validação cruzada e do conjunto de teste

```python

from pathlib import Path
import joblib
import sklearn

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier

# Pipeline com escalonamento e classificador
modelo_final = Pipeline(steps=[
    ("scaler", StandardScaler()),
    ("mlp", MLPClassifier(
        hidden_layer_sizes=(50,),
        activation="relu",
        solver="sgd",
        batch_size=64,
        learning_rate_init=0.01,
        n_iter_no_change=20,
        max_iter=200,
        random_state=42
    ))
])

# Treinamento com todos os dados disponíveis
modelo_final.fit(X, y)

# Criação da pasta e salvamento do modelo
pasta_modelos = Path("modelos")
pasta_modelos.mkdir(exist_ok=True)

caminho_modelo = pasta_modelos / "classificador_feijao.joblib"

joblib.dump(modelo_final, caminho_modelo)

print("Modelo treinado e salvo com sucesso!")
print("Caminho:", caminho_modelo)
print("Versão do scikit-learn:", sklearn.__version__)
print("Quantidade de amostras utilizadas:", len(X))
```

```python

# Rotina para recuperar o classificador salvo
def carregar_classificador(caminho="modelos/classificador_feijao.joblib"):
    return joblib.load(caminho)


# Recuperação do modelo
modelo_carregado = carregar_classificador()

print("Modelo recuperado com sucesso!")

# Previsões utilizando o modelo recuperado
y_pred_todos = modelo_carregado.predict(X)

print("Quantidade de previsões realizadas:", len(y_pred_todos))
```

```python
# Classes reconhecidas pelo classificador
classes = modelo_carregado.named_steps["mlp"].classes_

# Matriz de confusão
matriz = confusion_matrix(
    y,
    y_pred_todos,
    labels=classes
)

# Exibição gráfica
disp = ConfusionMatrixDisplay(
    confusion_matrix=matriz,
    display_labels=classes
)

fig, ax = plt.subplots(figsize=(10, 8))

disp.plot(
    ax=ax,
    xticks_rotation=45,
    values_format="d",
    colorbar=False
)

ax.set_title("Matriz de Confusão - Classificador Final")
plt.tight_layout()
plt.show()
```
